```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
import numpy as np
import pandas as pd
import seaborn as sns
from matplotlib import style
style.use('dark_background')
```

```python
x,y = make_blobs(n_samples = 200 , n_features =  2  , centers =2  , cluster_std  =  1.5 , random_state  =  42)
x
```

```python
x.shape
```

```python
y.shape
```

```python
plt.figure(figsize = (7,7))
plt.scatter(x[:,0],x[:,1],lw = 0.1,c = y,cmap = 'autumn')
plt.ylabel('values of y')
try:
  plt.xlabel('values of x')
except:
  plt.title('Data Visualization')
# formula of stright line  = mx + b
x  =  np.linspace(-5,5)
list  = [(1,2),(1,3),(1,4),(1,5),(1,6)]
for m,b in list :
  y = m*x + b
  plt.plot(x,y)

```

```python
from sklearn.svm import SVC
svc  =  SVC(kernel = 'linear',C = 1)
svc.fit(x,y)
```

```python
def support_vector_machine(svc,features , labels):

  plt.scatter(features[:,0],features[:,1],c = labels,cmap = 'autumn')

  plt.xlabel('values of x')

  plt.ylabel('values of y')
  ax = plt.gca()
  x_lim = ax.get_xlim()
  y_lim  = ax.get_ylim()


  i = np.linspace(x_lim[0],x_lim[1],50)
  j = np.linspace(y_lim[0],y_lim[1],50)
  ii,jj = np.meshgrid(j,i)
  ij  = np.vstack([ii.ravel(),jj.ravel()]).T
  kk = svc.decision_function(ij).reshape(jj.shape)
  ax.contour(ii,jj,kk,levels = [-1,0,1],colors =['r','g','b'],linestyles = ['--','-','--'],linewidths = 3)
  ax.scatter(svc.support_vectors_[:,0],svc.support_vectors_[:,1],s = 200,linewidth = 4,edgecolor = 'g')
  ax.set_xlim(x_lim)
  ax.set_ylim(y_lim)




```

```python
support_vector_machine(svc,x,y)
```

```python
SVM  = SVC(kernel = 'rbf',C = 5)
SVM.fit(x,y)
```

```python
support_vector_machine(SVM,x,y)
```

```python
from sklearn.datasets import make_circles
xn,yn = make_circles(100,factor = 0.1 , noise  = 0.06,random_state = 42)
```

```python
plt.scatter(xn[:,0],xn[:,1] ,s = 50,marker = 'o',edgecolors = 'red',cmap = 'coolwarm', c = yn)
ax  =  plt.gca()
ax.get_xlim()
ax.get_ylim()
plt.xlabel('X axis',fontsize  = 20)
plt.ylabel('y axis',fontsize = 20)
plt.title('Plotting x ans y')
plt.grid()
plt.show()
```

```python
svm1 = SVC(kernel = 'rbf',C  = 15)
svm1.fit(xn,yn)
```

```python
support_vector_machine(svm1,xn,yn)
```
